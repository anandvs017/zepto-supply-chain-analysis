# Zepto DRHP — Data Extraction & Compilation
**Source:** Zepto UDRHP-1 filed with SEBI on June 8, 2026

All figures manually extracted from the official prospectus. This notebook builds the master dataset used in all subsequent analysis.

```python
import pandas as pd
import numpy as np

# ── Financial Performance (from DRHP financials section) ──
financials = pd.DataFrame({
    'fiscal_year':        ['FY23',  'FY24',  'FY25',   'FY26'],
    'revenue_cr':         [2025,    4454,    11110,    22624],
    'net_loss_cr':        [1272,    1249,    4700,     5905],
    'ad_revenue_cr':      [None,    49,      651,      1636],
    'nrv_cr':             [None,    5231.7,  None,     24815.5]
})

# ── Supply Chain Operations (from DRHP operational metrics) ──
operations = pd.DataFrame({
    'fiscal_year':             ['FY24',  'FY25',  'FY26',  'Q4_FY26'],
    'dark_stores':             [700,     1000,    1139,    1139],
    'orders_per_day_per_store':[1325,    1565,    1677,    2140],
    'avg_distance_per_order_km':[2.05,   None,    None,    None],
    'annual_users_cr':         [1.06,    3.83,    4.79,    None],
    'repeat_order_rate_pct':   [94.87,   None,    96.62,   None]
})

# ── Customer Acquisition Cost (from DRHP unit economics) ──
cac = pd.DataFrame({
    'period':                     ['FY25',  'FY26',  'Q4_FY26'],
    'digital_mktg_cost_per_order': [33.75,   4.31,    1.01]
})

print('Financials:')
print(financials.to_string(index=False))
print('\nOperations:')
print(operations.to_string(index=False))
print('\nCAC trend:')
print(cac.to_string(index=False))
```

## Derived metrics
Calculating supply chain efficiency indicators not directly stated in the DRHP.

```python
# Loss as % of revenue — measures burn efficiency
financials['loss_pct_of_revenue'] = (
    financials['net_loss_cr'] / financials['revenue_cr'] * 100
).round(1)

# Ad revenue as % of NRV — measures monetisation quality
financials['ad_rev_pct_nrv'] = (
    financials['ad_revenue_cr'] / financials['nrv_cr'] * 100
).round(2)

# Revenue per dark store per year (crore)
rev_per_store = pd.DataFrame({
    'fiscal_year': ['FY24', 'FY25', 'FY26'],
    'revenue_cr':  [4454,   11110,  22624],
    'dark_stores': [700,    1000,   1139]
})
rev_per_store['revenue_per_store_cr'] = (
    rev_per_store['revenue_cr'] / rev_per_store['dark_stores']
).round(2)

print('Loss efficiency:')
print(financials[['fiscal_year','revenue_cr','net_loss_cr','loss_pct_of_revenue']].to_string(index=False))
print('\nRevenue per dark store (Cr):')
print(rev_per_store[['fiscal_year','revenue_per_store_cr']].to_string(index=False))
print('\nAd revenue as % of NRV:')
print(financials[['fiscal_year','ad_rev_pct_nrv']].dropna().to_string(index=False))
```

## Save cleaned datasets

```python
financials.to_csv('../data/zepto_financials_clean.csv', index=False)
operations.to_csv('../data/zepto_operations_clean.csv', index=False)
cac.to_csv('../data/zepto_cac_trend.csv', index=False)
rev_per_store.to_csv('../data/zepto_revenue_per_store.csv', index=False)
print('All datasets saved.')
```
